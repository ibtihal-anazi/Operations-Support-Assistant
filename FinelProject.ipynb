```python
"""
==============================================================================
📋 PROJECT BRIEF | Operations Support Assistant (مساعد الدعم التشغيلي)
- Description: نظام ذكاء اصطناعي لفرز وتذاكر المشاكل التقنية لخدمات السجل التجاري بوزارة التجارة.
==============================================================================
"""

import time
import torch
import tiktoken
import matplotlib.pyplot as plt
from pydantic import BaseModel, Field
from transformers import AutoModelForCausalLM, AutoTokenizer, TextIteratorStreamer
from threading import Thread

# إعداد مظهر الرسم البياني
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# ==============================================================================
# 🚀 LAB 1: Tokenization, Cost Analysis & Dashboards
# [توضيح]: تجهيز النموذج، حساب عدد الرموز (Tokens)، وتوليد رسوم بيانية (Slim Charts) للتنبؤ وتحليل تذاكر الدعم.
# [ملاحظة]: يوجد رسم بياني لمساعدة القادة التقنين في اتخاذ القرار نحو أكثر المشاكل التقنية وأعلاها خطورة لتقديم الدعم العاجل.
# ==============================================================================
MODEL_ID = "LiquidAI/LFM2.5-350M"
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
model = AutoModelForCausalLM.from_pretrained(MODEL_ID, torch_dtype=torch.bfloat16)
model.eval()

assistant_name = "Operations Support Assistant"

user_inquiries = [
    "The commercial registry update request has been delayed for three days and the status is still pending.",
    "I cannot log in to my ministry account despite entering the correct password, and an error message appears.",
    "A system error 500 occurs when trying to attach the required documents to the service form."
]

print("🤖 Assistant:", assistant_name)
print("="*60)

print("\n--- [Lab 1: Multiple Inquiries Tokenization] ---")
for i, inquiry in enumerate(user_inquiries, 1):
    token_ids = tokenizer.encode(inquiry)
    print(f"📌 Inquiry #{i}: {inquiry}")
    print(f"📊 Number of tokens: {len(token_ids)}")
    print("-" * 60)

print("\n" + "="*20 + " 📊 GENERATING MODERN SLIM CHARTS (Lab 1) " + "="*20)

sample_inquiry = "The commercial registry update request has been delayed and"
top_n = 6

inputs = tokenizer(sample_inquiry, return_tensors="pt")
with torch.no_grad():
    logits = model(**inputs).logits[0, -1]
    probs = torch.softmax(logits, dim=-1)
    top = torch.topk(probs, top_n)
    words = [tokenizer.decode([idx]) for idx in top.indices]
    values = [p.item() * 100 for p in top.values]

print(f"💬 Context Analyzed: {sample_inquiry}")
for word, value in zip(words, values):
    print(f"   {word!r:<15} probability: {value:.2f}%")

# 📊 Chart 1: Next Token Probabilities
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.bar([repr(w) for w in words], values, color="#3498DB", width=0.35, edgecolor="none", alpha=0.9)
ax.set_title("Next Token Probabilities (Prediction)", fontsize=11, fontweight='bold', pad=12)
ax.set_ylabel("Probability (%)", fontsize=9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()

# 📊 Chart 2: Impacted Services Dashboard
services = ["Commercial Registry", "Portal Authentication", "Error 500"]
case_counts = [2, 1, 1]

fig, ax = plt.subplots(figsize=(7.5, 3.5))
ax.bar(services, case_counts, color="#9B59B6", width=0.35, edgecolor="none", alpha=0.9)
ax.set_title("Operations Dashboard: Impacted Services & Frequency", fontsize=11, fontweight='bold', pad=12)
ax.set_ylabel("Number of Cases", fontsize=9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()

# 📊 Chart 3: Ticket Severity Levels
priorities = ["Critical (Error 500)", "High (Login Fail)", "Medium (Registry Delay)"]
priority_counts = [1, 1, 2]

fig, ax = plt.subplots(figsize=(7.5, 3.5))
ax.barh(priorities, priority_counts, color="#E74C3C", height=0.35, edgecolor="none", alpha=0.9)
ax.set_title("Operations Dashboard: Ticket Severity Levels", fontsize=11, fontweight='bold', pad=12)
ax.set_xlabel("Number of Tickets", fontsize=9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()


# ==============================================================================
# 🚀 LAB 2: API Client Integration, Streaming & TTFT
# [توضيح]: دمج نموذج الـ API مع تقنية البث الحي (Streaming) وآلية إعادة المحاولة (Retry) لقياس زمن الاستجابة (TTFT).
# ==============================================================================
print("\n" + "="*50)
print("🚀 STARTING LAB 2: API Client Integration & Streaming")
print("="*50)

def safe_generate_stream(prompt_text, max_retries=3):
    for attempt in range(1, max_retries + 1):
        try:
            inputs = tokenizer(prompt_text, return_tensors="pt")
            streamer = TextIteratorStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

            generation_kwargs = dict(
                inputs,
                streamer=streamer,
                max_new_tokens=65,
                temperature=0.7,
                repetition_penalty=1.2
            )

            thread = Thread(target=model.generate, kwargs=generation_kwargs)
            start_time = time.time()
            thread.start()

            first_token_time = None
            generated_text = ""

            print(f"\n🤖 {assistant_name} Response (Lab 2 Streaming):")
            print("-" * 60)

            for new_text in streamer:
                if first_token_time is None:
                    first_token_time = time.time()
                    ttft = (first_token_time - start_time) * 1000

                print(new_text, end="", flush=True)
                generated_text += new_text

            print("\n" + "-" * 60)
            if first_token_time:
                print(f"⏱️ Time to First Token (TTFT): {ttft:.2f} ms")

            return generated_text

        except Exception as e:
            if attempt < max_retries:
                time.sleep(1)
            else:
                return None

sample_ticket = "User statement: The commercial registry update request has been delayed for three days. Provide a quick support SOP action:"
print(f"\n📌 Processing Support Ticket via Lab 2 API:")
response = safe_generate_stream(sample_ticket)


# ==============================================================================
# 🚀 LAB 3: Tool Calling & Schema Validation
# [توضيح]: استخدام Pydantic لهيكلة تذاكر الدعم والتحقق من صحة البيانات (Schema Validation).
# ==============================================================================
print("\n" + "="*50)
print("🚀 STARTING LAB 3: Tool Calling & Integration Testing")
print("="*50)

class SupportTicketSchema(BaseModel):
    ticket_id: str = Field(..., description="معرف التذكرة التقنية")
    service_name: str = Field(..., description="اسم الخدمة مثل السجل التجاري أو المصادقة")
    issue_description: str = Field(..., description="وصف المشكلة التقنية")
    priority_level: str = Field(..., description="مستوى الأولوية: Critical, High, Medium")

print("\n--- [Lab 3: Code 6 — Integration Testing] ---")

def test_support_tool_integration():
    test_data = {
        "ticket_id": "TCK-90812",
        "service_name": "Commercial Registry",
        "issue_description": "Update request delayed for three days",
        "priority_level": "Medium"
    }

    try:
        validated_ticket = SupportTicketSchema(**test_data)
        print("✅ Integration Test Passed: Tool Schema Validated Successfully!")
        print(f"📋 Verified Ticket Data: {validated_ticket.model_dump()}")
        return True
    except Exception as e:
        print(f"❌ Integration Test Failed: {e}")
        return False

test_support_tool_integration()


# ==============================================================================
# 🚀 LAB 4: Recursive Chunking Pipeline
# [توضيح]: تقسيم أدلة التشغيل القياسية (SOP Documents) إلى أجزاء متداخلة لضمان دقة الاسترجاع.
# ==============================================================================
print("\n" + "="*50)
print("🚀 STARTING LAB 4: Recursive Chunking Pipeline")
print("="*50)

print("\n--- [Lab 4: Code 3 — Recursive Chunking for SOP Documents] ---")

def recursive_text_chunking(text, chunk_size=100, overlap=20):
    words = text.split()
    chunks = []

    for i in range(0, len(words), chunk_size - overlap):
        chunk = " ".join(words[i:i + chunk_size])
        chunks.append(chunk)

    return chunks

sop_document = (
    "Ministry of Commerce Operations Guide: "
    "1. Commercial Registry Updates: Verify customer submission status, check pending workflow approvals, "
    "and ensure all attachments comply with regulatory requirements before escalating. "
    "2. Portal Authentication Issues: Guide users to clear browser cache, verify credentials via national single sign-on, "
    "and reset passwords if authentication fails. "
    "3. Error 500 Handling: Capture error logs, verify API gateway status, and notify the Tier 2 technical team immediately."
)

chunked_docs = recursive_text_chunking(sop_document, chunk_size=35, overlap=10)

print(f"📄 Original SOP Document Length (Characters): {len(sop_document)}")
print(f"🧩 Total Chunks Generated: {len(chunked_docs)}")
print("-" * 60)

for idx, chunk in enumerate(chunked_docs, 1):
    print(f"🔹 Chunk #{idx}: {chunk}")
    print(f"📊 Chunk Token/Word Count: {len(chunk.split())} words")
    print("-" * 40)


# ==============================================================================
# 🚀 LAB 5: RAG Service Execution
# [توضيح]: ربط استعلامات المستخدم بالأجزاء المسترجعة (Context) لتوليد إجابات دقيقة مبنية على السياسات.
# ==============================================================================
print("\n" + "="*50)
print("🚀 STARTING LAB 5: RAG Service & Retrieval")
print("="*50)

print("\n--- [Lab 5: Code 2 — RAG Service Execution] ---")

def rag_support_service(query, chunks):
    relevant_context = chunks[0] if chunks else "No context available."

    rag_prompt = (
        f"Context Guidelines: {relevant_context}\n"
        f"User Support Inquiry: {query}\n"
        f"Provide a precise SOP action based on the context:"
    )

    print(f"🔍 Retrieved Context for Query: {query}")
    print(f"📦 Context Snippet: {relevant_context[:220]}...")

    return safe_generate_stream(rag_prompt)

query_sample = "How should I handle a delayed commercial registry update request?"
rag_response = rag_support_service(query_sample, chunked_docs)


# ==============================================================================
# 🚀 LAB 6: Conversation Memory (Sliding Window)
# [توضيح]: إدارة سياق المحادثة عبر نافذة تمرير (Sliding Window) للحفاظ على تسلسل الحوار بكفاءة.
# ==============================================================================
print("\n" + "="*50)
print("🚀 STARTING LAB 6: Conversation Memory (Sliding Window)")
print("="*50)

print("\n--- [Lab 6: Code 2 — Sliding Window Memory] ---")

class SlidingWindowMemory:
    def __init__(self, window_size=3):
        self.window_size = window_size
        self.history = []

    def add_message(self, role, message):
        self.history.append({"role": role, "message": message})
        if len(self.history) > self.window_size * 2:
            self.history = self.history[-self.window_size * 2:]

    def get_formatted_history(self):
        formatted = ""
        for entry in self.history:
            formatted += f"{entry['role']}: {entry['message']}\n"
        return formatted

memory = SlidingWindowMemory(window_size=2)
memory.add_message("User", "The commercial registry request is still pending.")
memory.add_message("Assistant", "Please verify customer submission status and workflow approvals.")

print(f"🪟 Active Sliding Window History (Size: {memory.window_size} turns):")
print("-" * 60)
print(memory.get_formatted_history())
print("-" * 60)


# ==============================================================================
# 🚀 LAB 7: Loop Detection & Execution Tracing
# [توضيح]: تطبيق نظام كشف التكرار (Loop Detector) وتتبع خطوات التنفيذ (Tracing) لضمان موثوقية النظام.
# ==============================================================================
print("\n" + "="*50)
print("🚀 STARTING LAB 7: Loop Detection & Tracing")
print("="*50)

# 1. Code 2: LoopDetector
print("\n--- [Lab 7: Code 2 — Loop Detector] ---")

class LoopDetector:
    def __init__(self, max_similar_responses=2):
        self.seen_responses = []
        self.max_similar = max_similar_responses

    def is_looping(self, response_text):
        count = self.seen_responses.count(response_text)
        self.seen_responses.append(response_text)
        return count >= self.max_similar

loop_detector = LoopDetector()
test_response = "Verify customer submission status in commercial registry portal."
loop_detector.is_looping(test_response)
is_repeating = loop_detector.is_looping(test_response)

print(f"🛡️ Loop Detection Status: {'⚠️ Loop Detected! Halting Generation.' if is_repeating else '✅ Pass — No Loop Detected.'}")

# 2. Code 3: Tracing & Execution Flow
print("\n--- [Lab 7: Code 3 — Execution Tracing] ---")

def trace_execution_step(step_name, status="Success", details=""):
    timestamp = time.strftime("%H:%M:%S")
    print(f"⏱️ [{timestamp}] TRACE: Step '{step_name}' -> Status: {status} | Details: {details}")

trace_execution_step("Ticket Ingestion", "Success", "Parsed 1 ticket")
trace_execution_step("Context Retrieval", "Success", "Retrieved 1 relevant SOP chunk")
trace_execution_step("LLM Inference", "Success", "Generated response with 65 tokens max")
trace_execution_step("Loop Check", "Pass", "No infinite loop detected")

print("\n" + "="*60)
print("🎉 Project Ended")
print("="*60)
```
